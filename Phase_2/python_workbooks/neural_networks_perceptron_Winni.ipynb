### 1c — Weight Update Rule

We update weights using gradient descent. For each sample $i$ and each weight $j$:

$$w_j^{\text{new}} = w_j^{\text{old}} + \eta \cdot (y_i - \hat{y}_i) \cdot x_{ij}$$

Where:
- $\eta$ = learning rate
- $y_i$ = true label
- $\hat{y}_i$ = predicted label
- $x_{ij}$ = the $j$-th feature of sample $i$

```python
def plot_decision_boundary(X, y, W, title="Decision Boundary"):
    """Plots the dataset and the perceptron's learned decision boundary."""
    fig, ax = plt.subplots(figsize=(6, 5))
    
    colors = ['cornflowerblue', 'tomato']
    labels = ['No Insomnia (0)', 'Insomnia (1)']
    for cls in [0, 1]:
        mask = y == cls
        ax.scatter(X[mask, 1], X[mask, 2], color=colors[cls], label=labels[cls],
                   s=120, edgecolors='black', zorder=3)
    
    # Decision boundary: W[0] + W[1]*x1 + W[2]*x2 = 0  =>  x2 = -(W[0] + W[1]*x1) / W[2]
    if W[2] != 0:
        x1_range = np.linspace(0, 5, 100)
        x2_boundary = -(W[0] + W[1] * x1_range) / W[2]
        ax.plot(x1_range, x2_boundary, 'k--', linewidth=2, label='Decision Boundary')
    
    ax.set_xlabel("Hours of Exercise / Week")
    ax.set_ylabel("Cups of Coffee / Week")
    ax.set_title(title)
    ax.legend()
    ax.set_xlim(0, 5)
    ax.set_ylim(0, 8)
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

plot_decision_boundary(X_insomnia, y_insomnia, W_init, title="Epoch 0 (Random Weights)")
plot_decision_boundary(X_insomnia, y_insomnia, W_after_1, title="After Epoch 1")
plot_decision_boundary(X_insomnia, y_insomnia, W_final, title=f"After 10 Epochs (lr=0.1)")
```

## Part 4 — Bonus: The Limits of a Single Perceptron

The slide deck ends with this key limitation: *a single-layer perceptron can only model **linearly separable** datasets*. Let's see this in action by adding the third Iris class (Virginica) back in.

**Run the cell below and observe what happens.**

```python
# Three-class Iris (no longer binary)
X_full = iris.data
y_full = iris.target

X_tr, X_te, y_tr, y_te = train_test_split(X_full, y_full, test_size=0.2, random_state=42)
scaler_full = StandardScaler()
X_tr_s = scaler_full.fit_transform(X_tr)
X_te_s = scaler_full.transform(X_te)

clf_full = Perceptron(max_iter=100, eta0=0.1, random_state=42)
clf_full.fit(X_tr_s, y_tr)
y_pred_full = clf_full.predict(X_te_s)

print("=== 3-Class Iris (Perceptron) ===")
print(classification_report(y_te, y_pred_full, target_names=iris.target_names))

print("\n💡 Notice that accuracy dropped. Why?")
print("   Versicolor and Virginica are NOT linearly separable.")
print("   This is exactly why we need a NETWORK of neurons, not just one perceptron!")
```
